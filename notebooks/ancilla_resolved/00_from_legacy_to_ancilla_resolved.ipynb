# From legacy TFIM to ancilla-resolved TFIM

A compatibility bridge: the historic API remains available while the new extension makes the ancilla record explicit.

```python
from gauge_simulation import models
from gauge_simulation.ancilla_resolved import ITEProtocol, TFIMSpec, clean_ground_energy
from gauge_simulation.ancilla_resolved.circuits import build_tfim_circuit

# Historical API: unchanged for the published notebooks.
legacy_model = models.build_hamiltonian("tfim", [1, 2], trotter_steps=1)
print(f"Legacy exact ground energy: {legacy_model.exact_ground_state():.8f}")

# Explicit extension: the same uniform clean TFIM specification.
spec = TFIMSpec.rectangular(rows=1, cols=2)
protocol = ITEProtocol(
    beta=0.75,
    steps=1,
    angle_mapping="linear",       # historical angle convention
    architecture="coherent_reuse", # historical ancilla schedule
    input_mode="purification",    # historical input preparation
)
bundle = build_tfim_circuit(spec, protocol)
print(f"Extension exact ground energy: {clean_ground_energy(spec):.8f}")
print(f"Extension circuit qubits: {bundle.circuit.num_qubits}")
print(f"Interpretation: {bundle.interpretation}")
```
